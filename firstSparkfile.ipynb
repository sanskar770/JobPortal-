```python
a=spark.sparkContext.textFile("Pysparktest1.txt")
```

```python
a.collect()
```

```python
b=a.flatMap(lambda x: x.split(" "))
b.collect()
```

```python
c=b.map(lambda x:(x,1))
c.collect()
```

```python
d=c.reduceByKey(lambda a,b:a+b)
d.collect()
```
